# K-Nearest Neighbors (KNN)



## Q1. What is K-Nearest Neighbors, and how does it classify a new point?


- KNN is a **classification algorithm** that works by comparing a new data point to existing labeled data.
- To classify a new sample:
  1. Compute the **distance** (commonly Euclidean) between the new point and all training points.
  2. Select the **k nearest neighbors** (the closest training samples).
  3. Assign the class label based on the **majority vote** among those neighbors.

For example, if \(k=5\) and among the 5 closest points, 3 are labeled "7" and 2 are labeled "9", the new point will be classified as "7".

### Why KNN is a "Lazy Learner"
- Unlike algorithms such as **Logistic Regression, SVM, or Decision Trees**, KNN does **not** build a model during training.
- During .fit():
  - KNN simply **stores the training data**.
  - No parameters are learned, no optimization is performed.
- The **real computational work happens during prediction (.predict())**:
  - Distances to all training points are calculated.
  - Neighbors are selected and majority voting is performed.

This is why KNN is called:
- **Lazy learner** → it postpones computation until prediction time.
- **Instance-based method** → it relies directly on stored instances rather than a learned model.

### Comparison with Other Algorithms
- **Logistic Regression / SVM / Decision Tree**:
  - Perform heavy computation during .fit() (optimization, splitting, parameter estimation).
  - Predictions are fast because the model is already built.
- **KNN**:
  - .fit() is trivial (just storing data).
  - Predictions are computationally expensive, especially with large datasets, because distances must be computed for every query.


In the **Digits dataset (1,797 samples, 64 features)**, KNN will store all pixel values during .fit(). When predicting a new handwritten digit, it compares the new 8×8 image to all stored images, finds the closest ones, and decides the digit based on majority vote.

### Why Feature Scaling is Critical in KNN?

- KNN classifies based on distance metrics (commonly Euclidean distance).

- If one feature (say, pixel intensity at a certain position) has a much larger numeric range than others:

  - Distances will be dominated by that feature.

  - The algorithm will effectively ignore smaller-range features, even if they are more informative.

- Example:

  - Suppose one pixel ranges from 0–1000, while others range from 0–16.

  - The large-range pixel will overshadow all others in the distance calculation.

- Scaling ensures all features contribute equally to the distance, preventing bias.

---

**Without scaling** → KNN might classify based on a single dominant pixel.

**With scaling** → Each pixel intensity is normalized, so the algorithm considers the overall pattern of the digit image, not just one feature.
